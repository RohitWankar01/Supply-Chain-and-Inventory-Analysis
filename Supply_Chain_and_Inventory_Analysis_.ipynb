### Step 3: Exploratory Data Analysis (EDA)

```python
#Q1 Which product types generate the highest revenue?
rev_product = df.groupby("Product type")["Revenue generated"].sum().sort_values(ascending=False)
display(rev_product)

plt.figure(figsize=(10,5))
rev_product.plot(kind="bar",color="red",alpha=0.7)
plt.title("Total Revenue by Product Type.")
plt.xlabel("Product Type")
plt.ylabel("Total revenue")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
```

### Analysis:
1. Skincare generates the highest revenue, approximately 242,000.
2. Cosmetics generates the lowest revenue, approximately 161,000.
### Conclusion:
1. Skincare is the top revenue-generating product category.
2. The business should analyze skincare sales to understand what drives its strong performance.

```python
#Q2. How do manufacturing costs vary across product types?
plt.figure(figsize=(10,5))
sns.boxplot(
    data= df,
    x= "Product type",
    y= "Manufacturing costs",

)
plt.title('Manufacturing Costs by Product Type')
plt.xlabel('Product Type')
plt.ylabel('Manufacturing Costs')

plt.show()
```

#### Analysis
1. Skincare has the widest variation in manufacturing costs, while haircare has a narrower distribution.
2. The median manufacturing costs are similar across all three product types, around 45–47.
#### Conclusion
1. Manufacturing costs vary more for skincare than for haircare.
2. The business should investigate why skincare manufacturing costs fluctuate more.

```python
#Q3. Is there a relationship between price and revenue generated?
plt.figure(figsize=(8, 5))
plt.scatter(
    df['Price'],
    df['Revenue generated'],

)
plt.title('Price vs Revenue Generated')
plt.xlabel('Price')
plt.ylabel('Revenue Generated')
plt.show()
```

#### Analysis
1. The points are widely scattered, showing no clear relationship between price and revenue.
2. Both low-priced and high-priced products generate different revenue levels.
#### Conclusion
1. Product price alone does not appear to explain revenue generation.
2. Other factors, such as sales volume, may also influence revenue.

```python
# Q4. Which transportation modes have the highest average shipping costs?
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='Transportation modes',
    y='Shipping costs'
)

plt.title('Average Shipping Costs by Transportation Mode')
plt.xlabel('Transportation Mode')
plt.ylabel('Average Shipping Costs')
plt.xticks(rotation=30)
plt.grid(axis="y")
plt.show()
```

#### Analysis
1. Air has the highest average shipping cost, approximately 6.0.
2. Sea has the lowest average shipping cost, approximately 5.0.
#### Conclusion
1. Air transportation is the most expensive option on average.
2. Sea transportation may help reduce shipping costs when delivery time permits.

```python
# Q5. Which supplier has the longest delivery time?
sns.barplot(
    data=df,
    x="Supplier name",
    y="Lead times",
    color="green",
    alpha=0.4
)
plt.title("Average Lead Time by Supplier")
plt.xlabel("Supplier name")
plt.ylabel("lead times")
plt.grid(axis="y")
plt.show()
```

#### Analysis
1. Supplier 4 has the longest average lead time, approximately 17.
2. Supplier 3 has the shortest average lead time, approximately 14.5.
#### Conclusion
1. Supplier 4 may require closer monitoring to improve delivery speed.
2. Supplier 3 has the shortest average lead time among the suppliers shown.

```python
#Q6. How are order quantities distributed?
plt.hist(
     df['Order quantities'],
     bins=10,
     edgecolor="black"
 )
plt.title("Distribution of quantites")
plt.xlabel("Order quantity")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()
```

#### Analysis
1. Order quantities are spread across a wide range, from small orders to large orders.
2. Higher frequencies appear around the 20–30, 50–70, and 75–85 quantity ranges.
#### Conclusion
1. The business handles orders of varying quantities rather than one consistent order size.
2. Understanding these patterns can help with inventory and order planning.

```python
# Q7. Is there a relationship between stock levels and the number of products sold?

plt.figure(figsize=(8, 5))
plt.scatter(
    df["Stock levels"],
    df["Number of products sold"]
)

plt.title("Stock Levels vs Number of Products Sold")
plt.xlabel("Stock Levels")
plt.ylabel("Number of Products Sold")
plt.grid(alpha=0.3)
plt.show()
```

## Analysis:
- The scatter plot shows no clear relationship between stock levels and the number of products sold.
-Products with higher stock levels do not consistently show higher sales.

## Conclusion:
- Stock levels alone may not explain sales performance. Inventory decisions should also consider demand patterns and   other relevant factors.

```python
#Q8 Which routes are used most frequently?
route_counts = df["Routes"].value_counts()
plt.pie(
    route_counts,
    labels=route_counts.index,
    autopct="%1.1f%%",                ## to show value in percentage

)
plt.title("Distribution of suppl chain routes")
plt.show()
```

### Analysis
1. Route A is used most frequently, followed closely by Route B.
2. Route C has the lowest share of records.
### Conclusion
1. Route A accounts for the largest proportion of supply chain route records.
2. The business can review route usage to improve transportation planning.

```python
#Q9. Which numerical variables are correlated with each other?
num_df = df.select_dtypes(include='number')
corr = num_df.corr()
plt.figure(figsize=(12,9))
sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=True
)
plt.title('Correlation Heatmap of Supply Chain Variables')

plt.show()
```

#### Analysis
1. Most numerical variables show weak correlations, with values close to zero.
2. Some variables show moderate relationships, but no strong correlation is clearly visible in the heatmap.
#### Conclusion
1. Most variables do not have a strong linear relationship with one another.
2. The business can investigate individual relationships further to identify factors affecting supply chain performance.
